## What is MuJoCo?

MuJoCo (*Multi-Joint dynamics with Contact*) is a physics simulator.
It simulates gravity, friction, and collisions so your code can be tested
safely before running on the real robot.

You don't need to know anything about MuJoCo — `SawyerSim` hides all of that.

## Step 1 — Import and create the simulation

```python
from sawyer_student import SawyerSim

# gui=True opens a 3-D viewer window.
# Set gui=False to run headlessly (e.g. on a server).
sim = SawyerSim(gui=True)
```

## Step 2 — Start the simulation

`sim.start()` compiles the robot description, builds the scene, and opens the viewer.
It also teleports the robot to a safe *ready pose* automatically.

> **Note:** This takes a few seconds the first time because it compiles URDF/MJCF files.

```python
sim.start()
```

## Step 3 — Check the robot's current position

The end-effector is the gripper at the tip of the robot arm.
Its position is given in metres relative to the world origin.

```python
x, y, z = sim.get_position()
print(f"End-effector position: x={x:.3f} m, y={y:.3f} m, z={z:.3f} m")
```

You can also get the full pose (position + orientation quaternion):

```python
pose = sim.get_pose()
print("Full pose:", pose)
```

## Step 4 — Move the robot

`move_to(x, y, z)` moves the end-effector to a target position.

- Coordinates are in **metres**, in the **world frame** (z=0 is the floor).
- The motion planner automatically finds joint angles and generates a smooth trajectory.
- Returns `True` if the motion succeeded, `False` if the position is unreachable.

The robot's workspace roughly spans:
- x: 0.2 – 0.7 m (in front of the robot)
- y: -0.4 – 0.4 m (left-right)
- z: 0.0 – 0.8 m (above the floor)

```python
# Move to a position in front of the robot, about 30 cm above the floor
success = sim.move_to(0.5, 0.0, 0.30)
print("Motion succeeded:", success)
```

```python
# Move to the left
sim.move_to(0.4, 0.2, 0.30)
```

```python
# Move to the right
sim.move_to(0.4, -0.2, 0.30)
```

### Speed control

Pass `speed="slow"`, `speed="medium"` (default), or `speed="fast"`.

```python
sim.move_to(0.5, 0.0, 0.30, speed="slow")
sim.move_to(0.5, 0.0, 0.30, speed="fast")
```

## Step 5 — Gripper control

```python
sim.open_gripper()
sim.wait(0.5)   # wait 0.5 s for the fingers to fully open

sim.close_gripper()
sim.wait(0.5)
```

## Step 6 — Reset and close

```python
# Return to the ready pose
sim.reset()
```

```python
# Shut down the simulation
sim.close()
```

---

## Summary

| Method | What it does |
|--------|-------------|
| `SawyerSim(gui=True)` | Create the simulation |
| `sim.start()` | Build scene and launch viewer |
| `sim.move_to(x, y, z)` | Move end-effector |
| `sim.open_gripper()` | Open fingers |
| `sim.close_gripper()` | Close fingers |
| `sim.get_position()` | Read end-effector (x, y, z) |
| `sim.wait(seconds)` | Pause |
| `sim.reset()` | Return to ready pose |
| `sim.close()` | Shut down |
