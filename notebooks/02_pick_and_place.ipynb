# 02 — Pick and Place

In this notebook you will:
1. Add a box to the simulation table
2. Move the robot to pick up the box
3. Move it to a new location and release it

This is the classic robotics task: **pick and place**.

---

## The strategy

A robust pick sequence always follows these steps:

```
1. Open gripper
2. Move ABOVE the object  (approach)
3. Move DOWN to the object  (descend)
4. Close gripper  (grasp)
5. Move UP  (lift)
6. Move to the place position  (transport)
7. Open gripper  (release)
8. Move up and away  (retreat)
```

Approaching from above prevents the arm from knocking the object over.

## Setup

````python
from sawyer_student import SawyerSim

sim = SawyerSim(gui=True)

# Place a 5 cm cube on the table surface (z=0)
# The cube centre is at z=0.025 (half its height)
BOX_X = 0.45
BOX_Y = 0.13
BOX_Z = 0.025  # half of 0.05 m height

sim.add_box(
    name="cube",
    position=[BOX_X, BOX_Y, BOX_Z],
    size=[0.05, 0.05, 0.05],
    color=[0.9, 0.3, 0.1, 1.0],  # reddish-orange
)

sim.start()
````

## Phase 1 — Open gripper and approach

First, open the gripper so we don't hit the cube when approaching.
Then move to a position **15 cm above** the cube.

````python
APPROACH_HEIGHT = 0.15  # 15 cm above the cube

print("Opening gripper…")
sim.open_gripper()
sim.wait(0.3)

print("Approaching from above…")
sim.move_to(BOX_X, BOX_Y, BOX_Z + APPROACH_HEIGHT)
sim.wait(0.2)
````

## Phase 2 — Descend to grasp height

Move down until the gripper fingers are around the cube.
We stop about 1 cm above the floor to avoid scraping.

````python
GRASP_HEIGHT_OFFSET = 0.015  # 1.5 cm above the cube centre

print("Descending to grasp position…")
sim.move_to(BOX_X, BOX_Y, BOX_Z + GRASP_HEIGHT_OFFSET, speed="slow")
sim.wait(0.3)
````

## Phase 3 — Grasp

````python
print("Closing gripper (grasping)…")
sim.close_gripper()
sim.wait(1.0)  # Give fingers time to close fully
````

## Phase 4 — Lift

````python
LIFT_HEIGHT = BOX_Z + 0.25  # 25 cm above the floor

print("Lifting…")
sim.move_to(BOX_X, BOX_Y, LIFT_HEIGHT)
sim.wait(0.3)
````

## Phase 5 — Transport to place position

````python
# Place the cube 20 cm to the right of where it was
PLACE_X = BOX_X
PLACE_Y = BOX_Y - 0.20
PLACE_Z = BOX_Z + 0.10  # slightly above the table surface

print(f"Transporting to place position ({PLACE_X:.2f}, {PLACE_Y:.2f})…")
sim.move_to(PLACE_X, PLACE_Y, LIFT_HEIGHT)  # keep height while moving sideways
sim.move_to(PLACE_X, PLACE_Y, PLACE_Z, speed="slow")  # descend
sim.wait(0.3)
````

## Phase 6 — Release

````python
print("Releasing (opening gripper)…")
sim.open_gripper()
sim.wait(1.0)
````

## Phase 7 — Retreat and reset

````python
print("Retreating…")
sim.move_to(PLACE_X, PLACE_Y, LIFT_HEIGHT)  # lift up before moving away

print("Returning to ready pose…")
sim.reset()
sim.wait(1.0)
````

````python
sim.close()
````
